```python
import pandas as pd

data = {
    'Name': ['Amit', 'Rahul', 'Priya', 'Neha', 'Vikas',
             'Anjali', 'Rohit', 'Sneha', 'Karan', 'Pooja'],
    'Age': [25, 30, 28, 35, 40, 26, 32, 29, 45, 31],
    'Department': ['IT', 'HR', 'IT', 'Sales', 'HR',
                   'Sales', 'IT', 'HR', 'Sales', 'IT'],
    'Salary': [35000, 45000, 55000, 30000, 60000,
               40000, 75000, 50000, 80000, 65000],
    'Experience': [1, 4, 3, 2, 8, 2, 6, 5, 10, 4],
    'Performance': [75, 60, 90, 55, 85, 70, 95, 80, 92, 88]
}

df = pd.DataFrame(data)
```

```python
df.head()
```

```python
df.query("Salary > 45000 and Experience > 3 and Performance > 80")
```

```python
con = [
    df['Salary'] >= 75000,
    df['Salary'] >= 65000,
    df['Salary'] >= 40000, 
]

levels = ['High','Mid','Low']

df['cat'] = np.select(con,levels,default=False)
```
