```python
data = [10, 12, 14, 15, 16, 18, 20, 22, 25, 100, 120]
Q1 = np.percentile(data,25)
Q3 = np.percentile(data,75)
IQR = Q3 - Q1
IQR
```

```python
upper = Q3 + (1.5*IQR)
lower = Q1 - (1.5*IQR)
print(upper)
print(lower)
```

```python
without_outliers = []
for i in data:
    if i >lower and i < upper:
        without_outliers.append(i)
without_outliers 
```

```python
df3 = pd.DataFrame({
    'Name': ['Amit', 'Rahul', 'Neha', 'Priya'],
    'Marks': [45, 85, 67, 92]
})
```

```python
import numpy as np

conditions = [
    df3['Marks'] >= 80,
    df3['Marks'] >= 60,
    df3['Marks'] >= 40
]
choices = ['High', 'Average', 'Low']

df3['Performance'] = np.select(conditions, choices, default='Fail')
```
