```python
marks = [85, 45, 92, 33, 76, 58, 29, 95, 67, 40]
# Count passed students (marks >= 40)
pas = []
fail = []
for i in marks:
    if i >= 40:
        pas.append(i)
    else:
        fail.append(i)
print(f'Pass Student: {len(pas)}')
print(f'Fail Student: {len(fail)}')

```

```python
sm = sum(marks)
lm = len(marks)
```
